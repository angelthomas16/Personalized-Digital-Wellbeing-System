```python
import numpy as np

# 7. IDs
df['user_id'] = df['user_id'].str.strip().str.replace('U-', 'U', regex=False)

# 1,2,4. addiction_level: restore "None", drop rows that can't be recovered
df['addiction_level'] = df['addiction_level'].replace({'No Addiction': 'None', 'Unknown': np.nan})
df.loc[df['addiction_level'].isna() & (df['addicted_label'] == 0), 'addiction_level'] = 'None'
df = df[df['addiction_level'].notna()]

# 5. label always derived from the level
df['addicted_label'] = df['addiction_level'].isin(['Moderate', 'Severe']).astype(int)

# 3. "Unknown" categories -> most common value
for c in ['gender', 'stress_level', 'academic_work_impact']:
    df[c] = df[c].replace('Unknown', np.nan)
    df[c] = df[c].fillna(df[c].mode()[0])

# 6. outliers: keep only valid ranges
rules = {'daily_screen_time_hours': (3, 12), 'social_media_hours': (0.5, 6),
         'gaming_hours': (0, 4), 'work_study_hours': (0.5, 6), 'sleep_hours': (4.5, 9),
         'notifications_per_day': (20, 250), 'app_opens_per_day': (15, 180),
         'weekend_screen_time': (3, 15)}
for c, (lo, hi) in rules.items():
    df = df[df[c].between(lo, hi)]

# 8. one row per user
df = df.drop_duplicates(subset='user_id', keep='first')

# 9. integer columns
df['notifications_per_day'] = df['notifications_per_day'].astype(int)
df['app_opens_per_day'] = df['app_opens_per_day'].astype(int)
```

```python

print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

### Issue with Addiction Labels

While cleaning the dataset, I found that two columns, `addiction_level` and `addicted_label`, have missing values. There are 965 missing values in `addiction_level` and 9 missing values in `addicted_label`.

I also checked both columns together and found that some values are not matching properly. So, I did not fill these missing values by guessing or using the other column.

For now, I kept both columns in the cleaned dataset because we have not finalized which column will be used for addiction prediction. During the modelling stage, we can decide how to handle these missing values and check the inconsistent labels before training the model.
