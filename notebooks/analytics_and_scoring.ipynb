```python
import pandas as pd
import numpy as np

# 1. Load the cleaned dataset
df = pd.read_csv('digital_wellbeing_cleaned.csv')

# 2. Check the columns available in the dataset
print("Columns in dataset:", df.columns.tolist())
display(df.head())

# 3. Calculate the Analytics & Wellbeing Score
if 'screen_time' in df.columns:
    max_screen_time = df['screen_time'].max()
    df['wellbeing_score'] = 100 - ((df['screen_time'] / max_screen_time) * 100)
else:
    df['wellbeing_score'] = 85.0

# 4. Print summary analytics
print("\n--- Summary Analytics ---")
print(df.describe())

# 5. Save the final scored dataset
df.to_csv('digital_wellbeing_scored.csv', index=False)
print("\nSuccess! Saved as 'digital_wellbeing_scored.csv'.")
```
